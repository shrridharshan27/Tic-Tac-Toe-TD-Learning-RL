### Aim
To develop a Reinforcement Learning (RL) based Tic-Tac-Toe agent using Temporal Difference (TD) learning and an $\epsilon$-greedy action selection strategy, allowing the agent to learn optimal gameplay through repeated self-play and continuously improve its decision-making over multiple training episodes.

```
import random
import numpy as np
import matplotlib.pyplot as plt
from tictactoe.game import TicTacToe
from tictactoe.agent import RLAgent
from tictactoe.train import train, evaluate
```

### 1. Training Agents via Self-Play
We train two autonomous agents ($X$ and $O$) against each other across multiple episode budgets: 1,000, 5,000, and 10,000 episodes.

```
episodes_list = [1000, 5000, 10000]
results = {}

for ep in episodes_list:
    print(f"Training agents for {ep} episodes...")
    agent_x, agent_o = train(episodes=ep)
    eval_res = evaluate(agent_x, agent_o, games=1000)
    results[ep] = {
        'x_win': eval_res['x_win_pct'],
        'o_win': eval_res['o_win_pct'],
        'draw': eval_res['draw_pct'],
        'brain_x': len(agent_x.V),
        'brain_o': len(agent_o.V),
        'agent_x': agent_x
    }
    print(f"  Done: X Wins={eval_res['x_win_pct']:.1f}%, O Wins={eval_res['o_win_pct']:.1f}%, Draws={eval_res['draw_pct']:.1f}%")
    print(f"  State Coverage: X={len(agent_x.V)} states, O={len(agent_o.V)} states\n")
```

### 2. Empirical Performance Visualization

```
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

x_wins = [results[ep]['x_win'] for ep in episodes_list]
o_wins = [results[ep]['o_win'] for ep in episodes_list]
draws = [results[ep]['draw'] for ep in episodes_list]

axes[0].plot(episodes_list, x_wins, marker='o', label='X Wins %', color='#2b5c8f', lw=2)
axes[0].plot(episodes_list, o_wins, marker='s', label='O Wins %', color='#d9534f', lw=2)
axes[0].plot(episodes_list, draws, marker='^', label='Draws %', color='#5cb85c', lw=2)
axes[0].set_title('Outcome Distribution vs Training Episodes', fontweight='bold')
axes[0].set_xlabel('Episodes')
axes[0].set_ylabel('Percentage (%)')
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].legend()

states_x = [results[ep]['brain_x'] for ep in episodes_list]
axes[1].plot(episodes_list, states_x, marker='D', color='#8a2be2', lw=2)
axes[1].set_title('State-Value Space Exploration (Agent X)', fontweight='bold')
axes[1].set_xlabel('Episodes')
axes[1].set_ylabel('Unique States')
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

### 3. Conclusion & Findings
The Reinforcement Learning-based Tic-Tac-Toe agent successfully learned optimal gameplay through self-play using Temporal Difference (TD) learning and the $\epsilon$-greedy strategy. As training episodes increased, the agent's performance improved significantly, resulting in higher win rates and broader state coverage.
